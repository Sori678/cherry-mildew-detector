```python
import os

def remove_non_image_files(data_dir):
    """
    Iterates through folders and removes any file that is not an image.
    """
    image_extensions = ('.png', '.jpg', '.jpeg')
    folders = os.listdir(data_dir)
    
    for folder in folders:
        files = os.listdir(os.path.join(data_dir, folder))
        i = 0
        non_image_files = 0
        for file in files:
            if not file.lower().endswith(image_extensions):
                file_path = os.path.join(data_dir, folder, file)
                os.remove(file_path)
                non_image_files += 1
            else:
                i += 1
        print(f"Folder: {folder} - Images: {i} - Non-images removed: {non_image_files}")

# Define the path to the extracted data
data_dir = 'inputs/cherry-leaves'
remove_non_image_files(data_dir)
```

The code for Split (in DataCollection.ipynb)

```python
import os
import shutil
import random

def split_train_validation_test_data(data_dir, train_set_ratio, validation_set_ratio, test_set_ratio):
    """
    Splits the images into train, validation, and test folders based on provided ratios.
    """
    if train_set_ratio + validation_set_ratio + test_set_ratio != 1.0:
        print("Error: Ratios must sum to 1.0")
        return

    # Create target directories: inputs/train, inputs/validation, inputs/test
    labels = os.listdir(data_dir)
    for folder in ['train', 'validation', 'test']:
        for label in labels:
            os.makedirs(os.path.join('inputs', folder, label), exist_ok=True)

    for label in labels:
        files = os.listdir(os.path.join(data_dir, label))
        random.shuffle(files)

        train_set_files_qty = int(len(files) * train_set_ratio)
        validation_set_files_qty = int(len(files) * validation_set_ratio)

        for count, file in enumerate(files):
            source_path = os.path.join(data_dir, label, file)
            if count < train_set_files_qty:
                destination = os.path.join('inputs', 'train', label, file)
            elif count < (train_set_files_qty + validation_set_files_qty):
                destination = os.path.join('inputs', 'validation', label, file)
            else:
                destination = os.path.join('inputs', 'test', label, file)
            
            shutil.move(source_path, destination)

    # Remove the now empty original directory
    shutil.rmtree(data_dir)
    print("Data split completed successfully!")

# Define the path and execute
data_dir = 'inputs/cherry-leaves'
split_train_validation_test_data(data_dir, 0.7, 0.1, 0.2)
```
